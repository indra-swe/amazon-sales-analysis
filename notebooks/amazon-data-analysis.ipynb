Which categories dominate?

```python
top_categories = df['category'].value_counts().head(10)

plt.figure(figsize=(12, 6))
sns.barplot(
    x=top_categories.values,
    y=top_categories.index
)

plt.title('Top Product Categories')
plt.xlabel('Number of Products')
plt.ylabel('Category')
plt.show()
```

```python
plt.savefig('../outputs/top_categories.png')
```

Rating Distribution

```python
plt.figure(figsize=(10,6))
sns.histplot(df['rating'], bins=20)
plt.title("Distribution of Product Ratings")
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.savefig('../outputs/rating_distribution.png')
```

Actual vs Discounted Price

```python
plt.Figure(figsize=(10,6))

sns.scatterplot(
    x=df.index,
    y=df['actual_price'],
    color='royalblue',
    marker='o',
    label='Actual Price',
    alpha=0.7
)

sns.scatterplot(
    x=df.index,
    y=df['discounted_price'],
    color='darkorange',
    marker='X',
    label='Discounted Price',

)

plt.title('Actual vs Discounted Price', fontsize=14, pad=15)
plt.xlabel('Product Index / Identifier', fontsize=12)
plt.ylabel('Price (₹)', fontsize=12)
plt.legend(loc='upper right')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.savefig('../outputs/price_comparison.png')
```

Correlation Heatmap

```python
numeric_df = df.select_dtypes(include=[np.number])

plt.figure(figsize=(12, 10))
sns.heatmap(
    numeric_df.corr(), 
    annot=True, 
    cmap='coolwarm'
)
plt.title('Correlation Heatmap of Numerical Features', fontsize=14, pad=15)
plt.show()
```

**Feature Engineering**

```python
#Converting rating count to numeric.

df['rating_count'] = (
    df['rating_count']
    .str.replace(',', '', regex=False)
)

df['rating_count'] = pd.to_numeric(
    df['rating_count'],
    errors='coerce'
)
```

```python
#Creating category encoding
from sklearn.preprocessing import LabelEncoder

# Handle NaN values in category column
df['category'] = df['category'].fillna('Unknown')

encoder = LabelEncoder()
df['category_encoded'] = encoder.fit_transform(df['category'])
```

```python
df['product_name_length'] = df['product_name'].str.len()
```

```python
df['review_length'] = df['review_content'].str.len()
```

```python
numeric_df = df.select_dtypes(include=['number'])
```

```python
numeric_df.head()
```

**Advanced Heatmap**

```python
#actual_price and discounted_price = 0.96 extremely high correlation (multicollinearity).

heatmap_df = numeric_df.drop(columns=['discounted_price'])
```

```python
plt.figure(figsize=(12, 10))

sns.heatmap(
    heatmap_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
)

plt.title('Advanced Correlation Heatmap of Numerical Features')
plt.show()
```
