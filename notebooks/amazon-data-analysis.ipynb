BUSINESS QUESTION 2:
Is Discounting Aggressive?

```python
#Creating discount categories
def discount_level(x):
    if x < 20:
        return 'Low Discount'
    elif x < 50:
        return 'Medium Discount'
    else:
        return 'High Discount'
    
df['discount_level'] = df['discount_percentage'].apply(discount_level)
```

```python
#Count Products
discount_counts = df['discount_level'].value_counts()

discount_counts
```

```python
#Visualization
plt.figure(figsize=(10,6))

sns.countplot(
    x='discount_level',
    data=df,
    palette='viridis',
    order=['Low Discount', 'Medium Discount', 'High Discount']
)

plt.title('Discount Aggressiveness Levels')
plt.savefig('../outputs/discount_aggressiv_levels.png')
plt.show()
```

**Business Interpretation:**

Amazon heavily depends on promotional pricing strategies.<br>
Possible business implications:

- intense competition
- price-sensitive customers
- aggressive sales campaigns

BUSINESS QUESTION 4:
Which Categories Receive Highest Discounts?

```python
#Category Discount Analysis

category_discount = (
    df.groupby('main_category')['discount_percentage']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
#Visualization

plt.figure(figsize=(12,6))

sns.barplot(
    x=category_discount.values,
    y=category_discount.index,
)
plt.title('Top categories by Average Discount Percentage')
plt.xlabel('Average Discount Percentage')
plt.savefig('../outputs/top_categories_by_discount.png')
plt.show()
```

**Business Interpretation:**<br>

- Home Improvement products receive the highest average discounts, suggesting highly competitive pricing strategies in this category.

- Computers & Accessories and Electronics also show significantly high discounts, indicating aggressive promotional activity in technology-related markets.

- Categories such as Office Products and Toys & Games receive relatively lower discounts, which may imply lower competition, stronger pricing power, or reduced promotional dependency.

- Overall, technology and utility-based categories appear to rely heavily on discount-driven sales strategies.

**BUSINESS QUESTION 5:** <br>
Do High Discounts Increase Popularity?

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x='discount_percentage',
    y='rating_count',
    data=df
)

plt.title('Discount Percentage vs Rating Count')
plt.xlabel('Discount Percentage')
plt.ylabel('Rating Count')
plt.savefig('../outputs/discount_vs_rating_count.png')
plt.show()
```

**Business Interpretation:**<br>

- Products with medium-to-high discount percentages tend to receive higher customer engagement, as reflected by larger rating counts.

- However, the relationship between discount percentage and popularity is not strongly linear, suggesting that factors beyond pricing also influence customer interest.

- Several highly discounted products still receive relatively low engagement, indicating that discounts alone are insufficient to drive popularity.

- Product quality, brand reputation, customer trust, and category demand likely play major roles in influencing customer interactions.
