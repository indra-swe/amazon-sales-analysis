Data Cleaning

```python
df.isnull().sum()
```

```python
mode_rating = df['rating_count'].mode()[0]
df['rating_count'] = df['rating_count'].fillna(mode_rating)
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df['discounted_price'] = (df['discounted_price']
.str.replace('₹', '', regex=False)
.str.replace(',', '', regex=False)
.astype(float))
```

```python
df['actual_price'] = (df['actual_price']
.str.replace('₹', '', regex=False)   
.str.replace(',', '', regex=False)
.astype(float))
```

```python
df['discount_percentage'] = (df['discount_percentage']
.str.replace('%', '', regex=False)
.astype(float))
```

df['discounted_price'] = pd.to_numeric(    df['discounted_price'].astype(str)      .str.replace('₹', '', regex=False)      .str.replace(',', '', regex=False),
    errors='coerce'
)

```python
df['rating'].unique()
```

```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
```

```python
df['rating'].unique()
```

```python
median_rating = df['rating'].median()
df['rating'] = df['rating'].fillna(median_rating)
```

```python
df['rating'].unique()
```

```python
df['category'].value_counts()
```

Save Cleaned Dataset

```python
df.to_csv('../data/amazon_cleaned.csv', index=False)
```
